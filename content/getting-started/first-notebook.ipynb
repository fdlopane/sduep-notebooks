## 4. Reading in a file

Everything so far has been the notebook talking to itself. Now it reads a real
file.

The cell below does three things. It fetches a toolkit called pandas, which is
what Python uses for anything shaped like a spreadsheet. It reads
`data/zones_15.csv` from the folder beside this notebook. And it reports how
many rows arrived and what the columns are called.

This is the slowest cell in the notebook, because pandas is a large toolkit and
it is being fetched now rather than earlier. Give it up to a minute before you
decide anything has gone wrong.

````python
import pandas as pd

zones = pd.read_csv("data/zones_15.csv", encoding="utf-8-sig")

print("Rows read:", len(zones))
print("Columns:", list(zones.columns))
````

Fifteen rows and five columns, and now the whole table is sitting in memory
under the name `zones`. Run the cell below to look at it.

The eastings and northings are British National Grid coordinates, in metres.
They are what the Ordnance Survey uses instead of latitude and longitude, and
you will do something with them at the end.

````python
zones
````

## 5. The one cell you change

Every notebook on this course has a cell like the one below, fenced top and
bottom by a line of dashes and labelled PARAMETERS. It holds the values you are
allowed to change, and in a model of any size it is the only cell you will ever
touch. Everything underneath it reads those values and does as it is told.

There are two values here. `LOCAL_AUTHORITY` picks one of the five councils in
the file, and `ZONE_ID` picks a single zone. Run the cell as it arrives, then
run the two cells below it and see what they produce.

After that, come back and change `LOCAL_AUTHORITY` from `"Gateshead"` to
`"North Tyneside"`. Keep the quotation marks, because they are what tells Python
this is a piece of text rather than a name it should go looking for. Then run
this cell again, and the two below it again, and watch the answers change.

````python
# ---------------------------------------------------------------------------
# PARAMETERS
# ---------------------------------------------------------------------------

LOCAL_AUTHORITY = "Gateshead"   # one of the five names in the local_authority
                                # column: Newcastle upon Tyne, Gateshead,
                                # North Tyneside, South Tyneside, Sunderland

ZONE_ID = "Z05"                 # a single zone, from Z01 to Z15

# ---------------------------------------------------------------------------
````

### The zones in one local authority

````python
chosen = zones[zones["local_authority"] == LOCAL_AUTHORITY]

if len(chosen) == 0:
    print("Nothing in the file has local_authority set to:", LOCAL_AUTHORITY)
    print("Check the spelling against these five values:")
    for name in sorted(zones["local_authority"].unique()):
        print("   ", name)
else:
    print(LOCAL_AUTHORITY, "holds", len(chosen), "of the fifteen zones:")
    for name in chosen["zone_name"]:
        print("   ", name)
````

Notice what happens if you mistype the name. You do not get a red block, because
asking for a council that is not in the file is a perfectly sensible question
with the answer "none". You get an empty result and a prompt, and only because
somebody wrote the cell to give you one. A notebook tells you nothing matched
if it has been built to; otherwise it says nothing at all and lets you carry on.

### One zone in detail

````python
one_zone = zones[zones["zone_id"] == ZONE_ID]

if len(one_zone) == 0:
    print("Nothing in the file has zone_id set to:", ZONE_ID)
    print("The file runs from Z01 to Z15.")
else:
    row = one_zone.iloc[0]
    print("Zone id:         ", row["zone_id"])
    print("Zone name:       ", row["zone_name"])
    print("Local authority: ", row["local_authority"])
    print("Easting:         ", row["easting"])
    print("Northing:        ", row["northing"])
````

## 7. Fifteen dots

One last cell, and this one draws rather than prints. It takes the easting and
northing of each zone and plots them as points, which is enough to produce a
rough map of Tyne and Wear without any mapping software involved at all.

You should be able to identify the coast on the right-hand side, the Tyne
running roughly west to east across the middle, and Sunderland down in the
bottom corner. Fifteen dots is a crude picture of five local authorities and
somewhere around 1.1 million people. But it is recognisably the right shape, and
that is the first check you can make on any spatial data you are handed: does it
land where the place actually is.

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 8))

ax.scatter(zones["easting"], zones["northing"], s=70, color="#C8102E", zorder=3)

# Zones on the eastern side get their label to the left of the dot, so that
# nothing runs off the edge of the picture.
middle = (zones["easting"].min() + zones["easting"].max()) / 2

for zone in zones.itertuples():
    eastern = zone.easting > middle
    ax.annotate(zone.zone_name,
                (zone.easting, zone.northing),
                xytext=(-9, -12) if eastern else (9, 6),
                textcoords="offset points",
                ha="right" if eastern else "left",
                fontsize=8)

ax.set_aspect("equal")
ax.margins(0.12)
ax.grid(True, color="#DDDDDD", zorder=0)
ax.set_xlabel("Easting, metres (British National Grid)")
ax.set_ylabel("Northing, metres (British National Grid)")
ax.set_title("The fifteen zones, placed by their grid coordinates")

plt.show()
````
