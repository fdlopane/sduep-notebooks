# Environment check

Run this before the module opens. It takes about a minute the first time and
tells you whether the tools this course uses will work on the machine you plan
to study on.

There is nothing to install and nothing to set up. Everything you need arrived
with this page.

**What to do**

1. Wait until the circle in the top right of this page is hollow rather than
   filled. A filled circle means the Python engine is still starting, and the
   first start is the slow one.
2. Click the cell below once to select it.
3. Press Shift and Enter together to run it.
4. Wait. The first run downloads the scientific libraries into your browser's
   cache, which on an ordinary connection takes a minute or two. Later runs are
   much quicker because the download is already cached.
5. Read the last line of the output.

If the last line says everything passed, you are ready and there is nothing
further to do. If anything failed, copy the whole output and send it to me with
a note about whether you were on a work machine or a personal one.

```python
import sys
import platform
import time

started = time.time()
results = []


def check(label, function):
    """Run one check and record whether it worked."""
    try:
        detail = function()
        results.append((label, True, detail))
        print(f"  OK      {label:34s} {detail}")
    except Exception as error:
        results.append((label, False, repr(error)))
        print(f"  FAILED  {label:34s} {type(error).__name__}: {error}")


print("Environment check")
print("=" * 62)
print()
print(f"Python:   {sys.version.split()[0]}")
print(f"Platform: {platform.machine()}")
print()
print("Libraries")
print("-" * 62)


def import_pandas():
    import pandas
    return f"version {pandas.__version__}"


def import_numpy():
    import numpy
    return f"version {numpy.__version__}"


def import_scipy():
    import scipy
    return f"version {scipy.__version__}"


def import_matplotlib():
    import matplotlib
    return f"version {matplotlib.__version__}"


check("import pandas", import_pandas)
check("import numpy", import_numpy)
check("import scipy", import_scipy)
check("import matplotlib", import_matplotlib)

print()
print("Files and arithmetic")
print("-" * 62)


def read_shipped_csv():
    import pandas as pd
    table = pd.read_csv("data/zones_15.csv", encoding="utf-8-sig")
    return f"{len(table)} rows, {len(table.columns)} columns"


def do_arithmetic():
    import numpy as np
    matrix = np.ones((15, 15)) * 2.0
    balanced = matrix / matrix.sum(axis=1, keepdims=True)
    return f"row totals all {balanced.sum(axis=1)[0]:.1f}"


def draw_a_chart():
    import matplotlib
    matplotlib.use("agg")
    import matplotlib.pyplot as plt
    figure, axes = plt.subplots(figsize=(2, 1))
    axes.plot([1, 2, 3], [2, 1, 3])
    plt.close(figure)
    return "figure created"


def write_and_read_back():
    import pandas as pd
    frame = pd.DataFrame({"zone": ["Z01", "Z02"], "trips": [120, 340]})
    frame.to_csv("check_output.csv", index=False, encoding="utf-8-sig")
    return f"{len(pd.read_csv('check_output.csv', encoding='utf-8-sig'))} rows read back"


check("read a shipped CSV file", read_shipped_csv)
check("run matrix arithmetic", do_arithmetic)
check("draw a chart", draw_a_chart)
check("write a file and read it back", write_and_read_back)

print()
print("=" * 62)
failures = [label for label, passed, _ in results if not passed]
elapsed = time.time() - started

if failures:
    print(f"RESULT: {len(failures)} of {len(results)} checks failed, in {elapsed:.0f} seconds.")
    print("Failed:", ", ".join(failures))
    print("Copy this whole output and send it to your module tutor.")
else:
    print(f"RESULT: all {len(results)} checks passed, in {elapsed:.0f} seconds.")
    print("Nothing further to do. See you when the module opens.")
```

## If it did not work

**Nothing happened when you pressed Shift and Enter, or the circle in the top
right stayed filled.** The Python engine did not start. Wait thirty seconds and
try again before concluding anything, because the first start is slow.

**The output stopped part-way through with a message about a network request.**
Something between your machine and this site is blocking the download. This is
the case worth catching early, and it is the reason this check exists. Send me
the message.

**A file check failed but the library checks passed.** The page uses a browser
feature called a service worker to let Python read files. Private or incognito
windows in some browsers switch that feature off. Try again in an ordinary
window.

**Everything failed on your work laptop and everything passed at home.** That is
useful information rather than a disaster, and I would rather know now than in
week three. Send me both results and we will sort out where you will do the
practical work.
